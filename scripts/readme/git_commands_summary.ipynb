## 会话中用到的 Git 指令汇总（解释 + 示例）
下面是本次会话中常用到的指令的简明汇总与示例：

- `git status`：查看工作区状态（未跟踪/已修改/已暂存文件）。
- `git branch`：列出本地分支；当前分支前有 `*`。
- `git rev-parse --abbrev-ref HEAD`：仅显示当前分支名。
- `git checkout -b <branch>`：基于当前状态创建并切换到新分支（会保留未提交更改）。
- `git checkout <branch>`：切换到已有分支。
- `git add .`：将更改加入暂存区。
- `git commit -m "说明"`：将暂存区内容提交到本地仓库。
- `git restore .`：恢复工作区到上次提交状态（会丢失未提交更改）。
- `git fetch origin`：从远程抓取更新（不合并）。示例：`git fetch origin Alphabet`。
- `git merge origin/Alphabet`：将远程分支合并到当前分支。
- `git pull`：等同于 `fetch`+`merge`（取决于配置）。
- `git push`：将本地提交推送到远程。示例：`git push origin my-local-changes`。
- `git log --oneline --decorate --graph -n 10`：简洁查看最近提交历史。
